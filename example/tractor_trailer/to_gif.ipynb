```python
import imageio.v2 as imageio  # ensure v2 API
import os

folder = "animation_buffer_t_shape_long_rollout"
output = "output.gif"

exts = (".png", ".jpg", ".jpeg")

# Collect candidate files
files = sorted([
    os.path.join(folder, f)
    for f in os.listdir(folder)
    if f.lower().endswith(exts)
])

print(f"Found {len(files)} candidate images.")

images = []
for f in files:
    try:
        # skip zero-byte files
        if os.path.getsize(f) == 0:
            print(f"Skipping empty file: {f}")
            continue

        img = imageio.imread(f)
        images.append(img)
        print(f"Loaded: {f}")
    except Exception as e:
        print(f"Skipping invalid image {f}: {e}")

if not images:
    raise RuntimeError("No valid images loaded. Check the folder and file types.")

# Save GIF (0.1 sec per frame)
imageio.mimsave(output, images[:210], duration=0.1)
print(f"GIF saved as {output}")
```
